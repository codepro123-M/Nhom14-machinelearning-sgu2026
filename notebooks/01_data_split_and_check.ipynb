```python
import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# 1. ĐỌC DỮ LIỆU GỐC
raw_data_path = "../Data/raw/bank-additional-full.csv"
df = pd.read_csv(raw_data_path, sep=";")
print(f"[*] Kích thước dữ liệu gốc: {df.shape[0]:,} dòng, {df.shape[1]} cột")

# 2. KIỂM TRA VÀ LOẠI BỎ DÒNG TRÙNG LẶP (DUPLICATES)
num_duplicates = df.duplicated().sum()
print(f"[*] Số dòng trùng lặp phát hiện: {num_duplicates} dòng")

if num_duplicates > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print(f"    -> Đã loại bỏ trùng lặp. Dữ liệu còn: {df.shape[0]:,} dòng")

# 3. KIỂM TRA GIÁ TRỊ 'unknown'
print("\n[*] Thống kê các cột có giá trị 'unknown':")
cat_cols = df.select_dtypes(include='object').columns.tolist()
for col in cat_cols:
    unknown_count = (df[col] == 'unknown').sum()
    if unknown_count > 0:
        percent = (unknown_count / len(df)) * 100
        print(f"    - Cột '{col:12}': {unknown_count:,} giá trị ({percent:.2f}%)")

# 4. KIỂM TRA PHÂN PHỐI NHÃN BIẾN MỤC TIÊU 'y'
print("\n[*] Phân phối biến mục tiêu (y):")
target_counts = df['y'].value_counts()
target_ratio = df['y'].value_counts(normalize=True) * 100
for label in target_counts.index:
    print(f"    - Nhãn '{label}': {target_counts[label]:,} mẫu ({target_ratio[label]:.2f}%)")

# 5. CHIA TẬP TRAIN / TEST (STRATIFIED, RANDOM_STATE=42, TEST_SIZE=0.2)
train_df, test_df = train_test_split(
    df, 
    test_size=0.2, 
    random_state=42, 
    stratify=df['y']
)

print(f"\n[*] Đã chia dữ liệu thành công:")
print(f"    - Tập Train: {train_df.shape[0]:,} dòng ({len(train_df)/len(df)*100:.1f}%)")
print(f"    - Tập Test : {test_df.shape[0]:,} dòng ({len(test_df)/len(df)*100:.1f}%)")

train_yes_pct = (train_df['y'] == 'yes').mean() * 100
test_yes_pct = (test_df['y'] == 'yes').mean() * 100
print(f"    -> Tỷ lệ 'yes' trên Train: {train_yes_pct:.2f}% | Test: {test_yes_pct:.2f}%")

# 6. TỰ ĐỘNG LƯU RA 2 FILE train.csv VÀ test.csv CHO NHÓM
processed_dir = "../Data/processed"
os.makedirs(processed_dir, exist_ok=True)

train_path = os.path.join(processed_dir, "train.csv")
test_path = os.path.join(processed_dir, "test.csv")

train_df.to_csv(train_path, sep=";", index=False)
test_df.to_csv(test_path, sep=";", index=False)

print(f"\n[*] XUẤT THÀNH CÔNG 2 FILE TẠI: {processed_dir}")
```
